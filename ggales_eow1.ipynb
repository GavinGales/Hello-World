```python
python_list = [1,2,3,4,5,6,7,8,9]
```

creating the python list

```python
import statistics as s
```

importing the statistics module

```python
list_sum = sum(python_list)
list_min = min(python_list)
list_max = max(python_list)
list_mean = s.mean(python_list)
list_median = s.median(python_list)
```

creating the variable for the stats

```python
print(f"Sum: {list_sum}")
print(f"Mean: {list_mean}")
print(f"Median: {list_median}")
print(f"Min: {list_min}")
print(f"Max: {list_max}")
```

```python
cases = 12
price_per_case = 34.99
discount_rate = 0.04
tax_rate = 0.06
```

Creating the variables of the constants 

```python
order_subtotal = cases * price_per_case
discount = order_subtotal * discount_rate
after_discount = order_subtotal - discount
tax = after_discount * tax_rate
order_total = after_discount + tax
```

doing the calculations and making them equal to the varibales

```python
print(f"Order subtotal: ${order_subtotal}")
print(f"Discount: ${discount}")
print(f"Tax: ${tax}")
print(f"Order total: ${order_total}")
```
